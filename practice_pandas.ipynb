```python

import pandas as pd

print("Pandas 불러오기 성공!")
print("Pandas 버전:", pd.__version__)
```

```python

# 가상의 반도체 식각률 데이터
data = {
    "Wafer": ["W01", "W02", "W03", "W04", "W05"],
    "Etch_Rate": [100, 105, 98, 102, 110]
}

# DataFrame 생성
df = pd.DataFrame(data)

# 데이터 출력
display(df)
```

```python

# 평균 식각률
print("평균:", df["Etch_Rate"].mean())

# 최대 식각률
print("최댓값:", df["Etch_Rate"].max())

# 최소 식각률
print("최솟값:", df["Etch_Rate"].min())

# 전체 데이터 개수
print("데이터 개수:", len(df))
```

```python

import matplotlib.pyplot as plt

plt.bar(df["Wafer"], df["Etch_Rate"])

plt.xlabel("Wafer")
plt.ylabel("Etch Rate (nm/min)")
plt.title("Etch Rate by Wafer")

plt.show()
```
